# **Evaluating Consumer Credit Risk: A Behavioral and Empirical Feature Engineering Approach**

To what extent can engineered financial and behavioral features improve the predictability of next-month credit card defaults?

```python
#2 Load the complete UCI Default of Credit Card Clients dataset and state the observation unit, target, prediction point, and target horizon.
```

**Observation unit and target**

* **Observation unit**: one credit-card client at the reporting snapshot.
* **Target**: default_next_month, where 1 means default on payment in the following month and 0 means no default in the following month.
* **Prediction point**: the reporting snapshot, before the following-month outcome is observed.
* **Target Horizon**: Next month
(t+1)

```python
# Load the dataset and the libraries and rename the target column
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

credit = pd.read_excel("default of credit card clients.xls", header=1)
credit = credit.rename(columns={'default payment next month': 'default_next_month'})
```

```python
#3 Audit data types, duplicates, missing values, zero denominators, target balance, and status-code ranges.
```

```python
print("Shape:", credit.shape)

def generate_data_audit(credit: pd.DataFrame) -> pd.DataFrame:
  """
  Generates a comprehensive data audit report for statistical
  and quality analysis of the columns in the credit DataFrame.
  """
  audit_dict = {
       'Data_Types': credit.dtypes.astype(str),
       'Missing_Values': credit.isna().sum(),
       'Duplicates': credit.duplicated().sum(),
       'Zero_Values': (credit == 0).sum(),
       'unique': credit.nunique(),
       'min': credit.min(numeric_only=True),
       'max': credit.max(numeric_only=True),
       }

  return pd.DataFrame(audit_dict).T

audit_report = generate_data_audit(credit)
display(audit_report)
```

```python
target_counts = credit['default_next_month'].value_counts()
print('Target Counts:\n', target_counts)

print('\nTarget Percentages:')
print(credit['default_next_month'].value_counts(normalize=True) * 100)
```

```python
#4 Construct at least: one level feature; one ratio feature; one count or severity feature; one change feature; one transformed monetary feature; one interaction feature.
```

```python
# Initialize the features DataFrame
features_df = pd.DataFrame(index=credit.index)
features_df['ID'] = credit['ID']

#1 Level Feature - Categorizing AGE into 4 behavioral levels
features_df['age_level'] = pd.cut(
    credit['AGE'],
    bins=[20, 30, 40, 50, 90],
    labels=['Young', 'Adult', 'Middle-Aged', 'Senior']
)

#2 Ratio Feature -

"""Credit Utilization Rate for the most recent month"""

features_df['utilization'] = credit['BILL_AMT1'] /credit['LIMIT_BAL']

"""Payment to Bill Ratio"""

bill_columns = ['BILL_AMT1', 'BILL_AMT2', 'BILL_AMT3', 'BILL_AMT4', 'BILL_AMT5', 'BILL_AMT6']
pay_amt_columns = ['PAY_AMT1', 'PAY_AMT2', 'PAY_AMT3', 'PAY_AMT4', 'PAY_AMT5', 'PAY_AMT6']

total_bill_6m = credit[bill_columns].sum(axis=1)
total_paid_6m = credit[pay_amt_columns].sum(axis=1)

features_df['payment_to_bill_6m'] = total_paid_6m / total_bill_6m


#3 Count Feature

"""Total number of months with delayed payments"""

pay_columns = ['PAY_0', 'PAY_2', 'PAY_3', 'PAY_4', 'PAY_5', 'PAY_6']
features_df['overdue_count_6m'] = (credit[pay_columns] > 0).sum(axis=1)

#4 Change Feature

"""The change in the account balance from Month 2 to the latest Month 1"""

features_df['bill_change_recent'] = credit['BILL_AMT1'] - credit['BILL_AMT2']
features_df['bill_change_early'] = credit['BILL_AMT5'] - credit['BILL_AMT6']


#5 Transformed Monetary Feature

features_df['log_limit_bal'] = np.log1p(credit['LIMIT_BAL'])

#6 Interaction Feature

"""Joint state: high_utilization AND recent_overdue"""

features_df['recent_overdue'] = (credit['PAY_0'] > 0).astype(int)
features_df['high_utilization'] = (features_df['utilization'] >= 0.80).astype(int)

features_df['stress_flag'] = features_df['high_utilization'] * features_df['recent_overdue']


display(features_df.head())
```

```python
#5 Produce at least three labelled plots or grouped summaries. Include counts in every rate comparison.
```

```python
"""Combine data"""

analysis_df = features_df.copy()
analysis_df['default_next_month'] = credit['default_next_month']

""" Generate the summary table for all 30,000 rows"""

stress_summary = analysis_df.groupby('stress_flag')['default_next_month'].agg(
    Total_Clients='count',
    Default_Count='sum',
    Default_Rate=lambda x: f"{(x.mean() * 100):.2f}%"
)
stress_summary.index = ['No Stress (0)', 'High Stress (1)']
display(stress_summary)
```

```python
"""Default Rate by Overdue History Count"""

print("\n" + "=" * 60)
print("GROUPED SUMMARY: DEFAULT RATE BY OVERDUE COUNT (6 MONTHS)")
print("=" * 60)
overdue_summary = analysis_df.groupby('overdue_count_6m')['default_next_month'].agg(
    Total_Clients='count',
    Default_Count='sum',
    Default_Rate=lambda x: f"{(x.mean() * 100):.2f}%"
)
display(overdue_summary)
```

```python
"""Boxplot του Utilization Rate σε σχέση με το Target"""

plot_df = features_df.copy()
plot_df['default_next_month'] = credit['default_next_month']

plt.figure(figsize=(6, 4))
sns.boxplot(x='default_next_month', y='utilization', data=plot_df, showfliers=False)
plt.title('Credit Utilization Rate vs Default Status')
plt.xlabel('Default Next Month')
plt.ylabel('Utilization Rate')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Detailed Feature Specifications

#### 1. `age_level`
*   **Missingness Policy:** No missing values expected based on the initial data audit (`Missing_Values = 0` for `AGE`). Any future unexpected null entries will be captured as an independent unlabelled category (`'Unknown'`) or float as `NaN`.
*   **Financial Interpretation:** Captures non-linear lifecycle risk. Younger cardholders (**'Young'**) typically lack extensive credit histories and stable income streams, demonstrating highly volatile payment behavior compared to career professionals (**'Adult'** / **'Middle-Aged'**) or retirees (**'Senior'**).

#### 2. `utilization`
*   **Missingness Policy:** The data audit confirmed that `LIMIT_BAL` contains zero occurrences of `0`. This guarantees that pure division is perfectly safe from structural `ZeroDivisionError` runtime failures.
*   **Financial Interpretation:** Measures a borrower's reliance on revolving credit and remaining capacity. A high utilization profile indicates heavy debt dependency and financial stretching, which correlates strongly with elevated default rates, as verified by the generated boxplot.

#### 3. `payment_to_bill_6m`
*   **Missingness Policy:** Zero-billed accounts over the 6-month period (`sum(BILL_AMT) == 0`) lead to division anomalies (`inf`). In accordance with the project safety rules, these macro-infinities are programmatically replaced with `np.nan` to accurately reflect a missing credit interaction footprint.
*   **Financial Interpretation:** Measures a client's aggregate repayment velocity. High ratios signify borrowers who aggressively clear their statement balances, whereas extremely low ratios reveal consumers making only minimum payments due to underlying cash-flow restrictions.

#### 4. `overdue_count_6m`
*   **Missingness Policy:** No missing values present in raw status features. (Note: `PAY_1` is naturally absent due to the source publisher's original naming architecture).
*   **Financial Interpretation:** Captures the chronic frequency of delinquency. As empirically validated by your grouped statistics report, a linear rise in this integer count signals severe repayment distress, serving as the strongest standalone predictor (climbing up to a **70.32% default rate** at a count of 6).

#### 5. `bill_change_recent` & `bill_change_early`
*   **Missingness Policy:** No missing values present in raw financial statement streams.
*   **Financial Interpretation:** Captures balance velocity dynamics. An escalating short-term trend (`bill_change_recent > 0`) implies rapid balance accumulation or an immediate spending shock. Conversely, the historical trend (`bill_change_early`) serves as an early warning signal, capturing the precise onset of financial distress months before a formal delinquency sequence starts.

#### 6. `log_limit_bal`
*   **Missingness Policy:** Complete data availability. The use of `np.log1p` inherently safeguards the function against potential zero bounds by executing \(\log(1 + x)\).
*   **Financial Interpretation:** Compresses the heavily right-skewed monetary scale of credit limits into a normalized distribution. This prevents massive credit limits (outliers) from dominating distance calculations or coefficients during training, ensuring balanced algorithm learning across both small and large accounts.

#### 7. `stress_flag`
*   **Missingness Policy:** Fully deterministic interaction feature; dynamically inherits missingness behavior directly from its core components (`utilization` and `PAY_0`).
*   **Financial Interpretation:** Isolates a highly critical, distressed sub-population where the consumer is simultaneously operating near max card capacity (\(\ge 80\%\)) and actively missing their most recent payment deadline. Your grouped matrix confirms its massive predictive power, isolating a sub-population with a **56.24% default rate**.

```python
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer

"""Data Leakage Demonstration via Global Imputation"""

mock_data = pd.DataFrame({
    'payment_to_bill_6m': [0.50, 0.20, np.nan, 0.10],
    'default_next_month': [0, 0, 1, 1]
})

print(mock_data[['payment_to_bill_6m', 'default_next_month']], "\n")

imputer = SimpleImputer(strategy='mean')
leaked_features = imputer.fit_transform(mock_data[['payment_to_bill_6m']])

X_train, X_test, y_train, y_test = train_test_split(
    leaked_features, mock_data['default_next_month'], test_size=0.5, random_state=42
)

print("X_train (Leaked):\n", X_train)
```

We created a small controlled dataset with a missing value (NaN) in the payment_to_bill_6m feature. We then mistakenly applied a global SimpleImputer to calculate and fill the column's mean value before splitting the data into training and testing sets.

By computing the mean across the entire dataset first, the missing value of the borrower who ends up in the training set (X_train) was replaced by 0.26666667. This specific value was mathematically calculated using information from the other observations that were later sequestered into the test set.

This process directly leaks future/unseen information from the test set into the training phase. It violates data independence, artificially inflates model evaluation metrics during validation, and causes the machine learning model to fail when deployed on genuine production data.

# Feature Interpretations: Association vs. Causality

A rigorous economic and statistical evaluation of each engineered feature, explicitly decoupling historical statistical correlations (Association) from true underlying behavioral triggers (Causality).

---

### 1. `age_level`
*   **Association:** Statistically, younger age groups (`Young`) often correlate with higher default counts. However, this is a proxy for unobserved structural factors (e.g., lower income stability, shorter credit history).
*   **Causality:** Age itself does not *cause* financial default. A person turning 25 does not automatically trigger an inability to pay. The true causal drivers are underlying economic variables like job market volatility and lower liquid savings.

### 2. `utilization`
*   **Association:** There is a strong, positive statistical link between high utilization and default probability, as visually proven by the upward shift in the generated boxplot.
*   **Causality:** High utilization does not directly *cause* a default; rather, it reflects a state of high debt dependency. The actual causality stems from a cash-flow breaking point—when a borrower's monthly debt obligations exceed their disposable income, forcing a default.

### 3. `payment_to_bill_6m`
*   **Association:** Low repayment ratios are highly correlated with imminent default risk, isolating borrowers who struggle to clear their balances over a rolling 6-month window.
*   **Causality:** Making a minimum payment does not mechanically *cause* a default. The causal link resides in underlying insolvency or structural budgeting deficits, where the borrower lacks the financial capacity to pay down principal debt, leaving them highly vulnerable to shocks.

### 4. `overdue_count_6m`
*   **Association:** This feature shows a powerful, near-linear statistical association with the target variable, where a maximum count of 6 corresponds to a severe **70.32% default rate**.
*   **Causality:** Missing a past payment does not physically *cause* a future month's default. Instead, a high overdue count is a behavioral symptom of an ongoing, unresolved financial crisis or systemic credit distress that continuously carries over into the next billing cycle.

### 5. `bill_change_recent` & `bill_change_early`
*   **Association:** Rapidly increasing statement balances are statistically linked to an elevated risk profile, signaling either sudden spending spikes or progressive debt accumulation.
*   **Causality:** A rising balance does not *cause* an account to fail. The causal factor is the sudden shock to the borrower’s debt-to-income ratio. If the balance velocity outpaces the borrower's income growth, it creates an unsustainable financial burden that eventually causes a payment break.

### 6. `log_limit_bal`
*   **Association:** Lower credit limits are often statistically associated with slightly higher default rates in retail portfolios, as they are frequently assigned to higher-risk or unseasoned profiles.
*   **Causality:** Modifying a credit limit scale using a logarithm has zero causal impact on a human borrower's real-world behavior. It is a purely mathematical transform designed to cause better convergence in machine learning algorithms by neutralizing the impact of large monetary outliers.

### 7. `stress_flag`
*   **Association:** This compound interaction feature isolates a highly specific, distressed sub-population, boasting a catastrophic **56.24% default rate**.
*   **Causality:** The flag itself is a mathematical construct and carries no direct causal weight. The true causal mechanism is the **simultaneous convergence** of two distinct risk states: the borrower has completely exhausted their revolving credit capacity ($\ge 80\%$) *at the exact same time* they experienced an immediate liquidity failure (`PAY_0 > 0`), stripping away any financial safety net.

#### 1. Why is `payment_to_bill_6m` not interchangeable with `utilization`?
*   **Answer:** They measure completely different dimensions of credit risk. `utilization` is a static, point-in-time structural metric that measures a borrower's current debt dependency and available capacity in a single snapshot. Conversely, `payment_to_bill_6m` is a dynamic, behavioral flow metric that aggregates repayment velocity and cash-flow consistency over a 6-month historical window. A borrower can structurally maintain high utilization while consistently exhibiting strong repayment velocity, or vice-versa.

#### 2. What information is lost when a continuous variable is replaced by a binary flag?
*   **Answer:** Replacing a continuous variable with a binary flag removes the internal variance, magnitude, and granular distribution of the data. For instance, when partitioning credit utilization at a $\ge 80\%$ threshold to create a flag, the model can no longer differentiate between a borrower at 81% capacity and one at 120% capacity. It reduces a rich spectrum of linear risk intensity into a crude step-function, discarding fine-grained information that could help capture subtle credit distress signals.

#### 3. Why must a rolling feature be computed within the entity and in time order?
*   **Answer:** Rolling features must be strictly grouped by the specific entity (e.g., `ID`) to prevent cross-contamination of financial data between unrelated borrowers. Furthermore, they must be computed in strict chronological time order (from oldest to newest) to preserve the temporal logic of the events and eliminate lookahead bias. Calculating rolling statistics out of chronological order would allow future data points to influence past metrics, resulting in artificial performance metrics during model development.

#### 4. When would a maximum be more appropriate than a count?
*   **Answer:** A maximum statistic (`max`) is more appropriate than a count when evaluating the **severity of historical distress** rather than its frequency. In credit risk, a borrower who misses a single payment deadline by 90 days (Severity/Max) typically poses a significantly higher loss-given-default (LGD) risk than a borrower who missed three separate payments by only 1 or 2 days (Frequency/Count). The maximum metric flags the worst-case boundary of risk exposure.

#### 5. Which feature definitions would change if the prediction point moved one month earlier?
*   **Answer:** All time-dependent lag structures would shift backwards by exactly one period ($t-1$). Specifically, current `utilization` would be forced to look at `BILL_AMT2 / LIMIT_BAL` (August balance) instead of `BILL_AMT1`. The `bill_change_recent` would shift from ($t_1 - t_2$) to ($t_2 - t_3$), and the historical rolling aggregations (`overdue_count_6m` and `payment_to_bill_6m`) would drop the September snapshot entirely and incorporate an older historical month (Month 7) to maintain their 6-month lookback window.
